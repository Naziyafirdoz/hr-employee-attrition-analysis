```python
df["Attrition_Flag"] = df["Attrition"].apply(
    lambda x: 1 if x == "Yes" else 0
)

df
```

#### FEATURE ENGINEERING

```python
def salary_band(x):
    if x < 5000:
        return "Low"
    elif x < 10000:
        return "Mid"
    else:
        return "High"

df["Salary Band"] = df["MonthlyIncome"].apply(salary_band)
df
```

```python
def exp_level(x):
    if x < 7:
        return "Junior"
    elif x < 10:
        return "Mid"
    else:
        return "Senior"

df["Experience Level"] = df["TotalWorkingYears"].apply(exp_level)
df       
```

```python
df["Risk Flag"] = df.apply(
    lambda row: "High Risk" if row["OverTime"] == "Yes" and row["JobSatisfaction"] < 3 else "Low Risk",
    axis = 1
)
df
```

```python
def age_group(x):
    if x < 30:
        return "Young"
    elif x < 45:
        return "Mid - Career"
    else:
        return "Experienced"

df["Age Group"] = df["Age"].apply(age_group)
df
```

#### BUSINESS INSIGHTS

##### Department attrition

R&D records the highest attrition count, suggesting burnout or limited growth opportunities in technical roles. The company should investigate workload distribution and introduce clear career progression paths in R&D to reduce turnover.

##### Overtime and attrition

Employees working overtime are significantly more likely to leave (attrition rate ~30% vs ~10% for non-overtime). This indicates work-life imbalance is a primary driver of attrition. The comapny should review workload distribution and enforce overtime limits to retain employees.

##### Age group and attrition

Young employees (under 30) show the highest attrition rate, suggesting dissatisfaction with compensation or limited early career development. Introducing mentorship programs and competitive entry-level salaries would help reduce turnover in this segment.

##### Salary and job role

Manager and Research Director roles earn significantly higher salaries than Sales Representatives and Lab Technicians. Salary disparity may be contributing to lower satisfaction in junior roles. Reviewing the compensation structure for entry-level positions could improve retention.

##### High risk employees

Employees flagged as High Risk (working overtime + low job satisfaction) represent a significant portion of the workforce. These employees are immediate attrition risks. HR should prioritize one-on-one reviews and intervention programs for this group before they exit.

##### Correlation insight from heatmap

Monthly income shows strong positive correlation with job level and total working years, meaning seniority drives pay more than performance or department. This rigid pay structure may demotivate high-performing junior employees. A performance-linked pay review is recommended.

# Final Conclusion

This HR Attrition Analysis project identified major drivers influencing employee attrition.

Key findings include:

- Research & Development department experienced the highest employee attrition.
- Employees working overtime showed significantly higher attrition rates.
- Younger employees demonstrated greater turnover tendency.
- Salary differences across job roles may influence employee retention.
- High-risk employees (overtime + low satisfaction) require proactive intervention.
- Compensation growth appears strongly related to job level and experience.

Recommendations:

1. Improve workload balance and reduce overtime.
2. Create career development programs for early-career employees.
3. Conduct employee engagement initiatives in high attrition departments.
4. Review compensation strategies for operational roles.
5. Monitor high-risk employees proactively.

This analysis demonstrates how data analytics can support HR decision-making and improve employee retention strategies.
